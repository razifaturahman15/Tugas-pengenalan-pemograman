```python
# ==========================================
# PROGRAM KASIR OTOMATIS
# Nama File: kasir.otomatis
# ==========================================

# Input data dari pengguna
print("=" * 36)
print("       PROGRAM KASIR OTOMATIS")
print("=" * 36)

nama_pelanggan = input("Masukkan Nama Pelanggan : ")
total_awal = float(input("Masukkan Total Belanja  : Rp "))
persentase_diskon = float(input("Masukkan Diskon (%)    : "))

# Proses perhitungan diskon
nominal_diskon = total_awal * (persentase_diskon / 100)

# Proses perhitungan total bayar
total_bayar = total_awal - nominal_diskon

# Format angka menjadi Rupiah
def format_rupiah(angka):
    return f"Rp{angka:,.2f}"

# ==========================================
# OUTPUT NOTA PEMBAYARAN
# ==========================================

lebar = 36

print("\n")
print("=" * lebar)
print(f"{'NOTA PEMBAYARAN':^{lebar}}")
print(f"{'KASIR OTOMATIS':^{lebar}}")
print("=" * lebar)

# Nama pelanggan rata tengah
print(f"{nama_pelanggan:^36}")
print("-" * lebar)

# Rincian pembayaran
print(f"{'Total Awal':<18}{format_rupiah(total_awal):>18}")
print(f"{f'Diskon ({persentase_diskon:g}%)':<18}{format_rupiah(nominal_diskon):>18}")
print("-" * lebar)
print(f"{'Total Bayar':<18}{format_rupiah(total_bayar):>18}")

print("=" * lebar)
print(f"{'TERIMA KASIH':^{lebar}}")
print(f"{'SELAMAT BERBELANJA KEMBALI':^{lebar}}")
print("=" * lebar)
```
